Simular dos variables aleatorias $X_1$ y $X_2$ que siguen distribuciones de Poisson con parametros $\mu_1 = 1$ y $\mu_2 = 2$ respectivamente. Calcular la variable $Y = X_1 + X_2$. Repetir la simulación 1000 veces. Graficar la distribución de la variable $Y$ y comparar con una distribución de probabilidad adecuada.

```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.stats import poisson
```

```python
#Simulo las variables aleatorias X1 y X2 que siguen una distribución de Poisson
mu1 = 1
mu2 = 2
X1 = poisson.rvs(mu1, size=1000)
X2 = poisson.rvs(mu2, size=1000)
Y = X1 + X2
```

### Suma de dos Poissionianas 
Como se explicó en el teórico, la suma de dos variables aleatorias $X_1$ y $X_2$ que siguen una distribución de Poisson nos devuelve una nueva variable aleatoria Y, que también sigue una distribución de Poisson, cuya media está dada por $E(Y) = E(X_1) + E(X_2) = \mu_1 + \mu_2$ y su varianza por $Var(Y) = Var(X_1) + Var(X_2) = \mu_1 + \mu_2$. Por esta razón se decidió comparar la distribución de la variable Y con una distribución de Poisson.

```python
k = np.arange(0,11)
mu = mu1 + mu2
bins = np.arange(-0.5,10.5,1)
poisson_prob = poisson.pmf(k, mu)
plt.hist(Y, bins, density = True, label = "Simulaciones")
plt.plot(k, poisson_prob, marker='o', linestyle= '--', color='tab:red', label='Distribución de Poisson')
plt.xlabel('Distribución de la variable Y')
plt.ylabel('Frecuencias')
plt.legend()
plt.show()
```
